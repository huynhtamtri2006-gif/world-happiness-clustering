```python
import os
import sys
import json
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập đường dẫn tương đối độc lập môi trường
project_root = Path('..').resolve()
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from src.models.clustering import match_clusters_greedy, euclidean_distance, run_kmeans
from src.features.feature_engineering import scale_features, SimplePCA, FEATURE_COLUMNS
from src.visualization.plots import plot_clusters_2d, plot_cluster_radar

tables_dir = project_root / 'reports' / 'tables'
figures_dir = project_root / 'reports' / 'figures'
processed_dir = project_root / 'data' / 'processed'
interim_path = project_root / 'data' / 'interim' / 'happiness_merged.csv'

tables_dir.mkdir(parents=True, exist_ok=True)
figures_dir.mkdir(parents=True, exist_ok=True)
processed_dir.mkdir(parents=True, exist_ok=True)

sns.set_theme(style='whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['figure.dpi'] = 300

# Định nghĩa nhãn định danh cụm bằng tiếng Việt theo chuẩn báo cáo
cluster_names = {
    0: 'Cụm 1 — Hạnh phúc cao (Phát triển toàn diện)',
    1: 'Cụm 2 — Hạnh phúc trung bình (Đang phát triển)',
    2: 'Cụm 3 — Hạnh phúc thấp (Khó khăn & Thách thức)'
}

years = [2015, 2016, 2017, 2018, 2019]
print('Môi trường phân tích và các thư viện đã sẵn sàng.')
```

## PHẦN A. PHÂN TÍCH PROFILE CỤM NĂM CƠ SỞ (2019)
Thực hiện phân tích chuyên sâu cho 156 quốc gia của năm 2019 theo đúng đề cương đồ án với số cụm tối ưu $k^* = 3$.

```python
# 1. Đọc dữ liệu năm 2019
df_all = pd.read_csv(interim_path)
df_2019 = df_all[df_all['year'] == 2019].dropna(subset=FEATURE_COLUMNS).copy().reset_index(drop=True)
print(f'Số lượng quốc gia năm 2019: {len(df_2019)}')

# 2. Chuẩn hóa đặc trưng và huấn luyện K-Means (k=3)
X_scaled, _ = scale_features(df_2019, method='standard')
labels_raw, model_2019 = run_kmeans(X_scaled, n_clusters=3, random_state=42)
df_2019['cluster_raw'] = labels_raw

# Chuẩn hóa thứ tự cụm theo mức điểm hạnh phúc thực tế giảm dần (0: Cao, 1: TB, 2: Thấp)
mean_scores = df_2019.groupby('cluster_raw')['happiness_score'].mean().to_dict()
sorted_clusters = sorted(mean_scores.keys(), key=lambda c: mean_scores[c], reverse=True)
mapping_2019 = {orig: canonical for canonical, orig in enumerate(sorted_clusters)}
df_2019['cluster'] = df_2019['cluster_raw'].map(mapping_2019)
df_2019['cluster_label'] = df_2019['cluster'].map(cluster_names)

# 3. Trực quan hóa không gian PCA 2D năm 2019
pca_2019 = SimplePCA(n_components=2)
X_pca_2019 = pca_2019.fit_transform(X_scaled)

plt.figure(figsize=(9, 6))
palette_2019 = {0: '#2ecc71', 1: '#f39c12', 2: '#e74c3c'}
for c_id in [0, 1, 2]:
    mask = (df_2019['cluster'] == c_id)
    plt.scatter(X_pca_2019[mask, 0], X_pca_2019[mask, 1], c=palette_2019[c_id], 
                label=cluster_names[c_id], s=65, alpha=0.85, edgecolors='none')

plt.title(f'Phân Bố 3 Cụm Quốc Gia Năm 2019 Trên Không Gian 2D PCA (n={len(df_2019)})', fontsize=13, pad=15, fontweight='bold')
plt.xlabel(f'PC1 ({pca_2019.explained_variance_ratio_[0]*100:.1f}%)', fontsize=11)
plt.ylabel(f'PC2 ({pca_2019.explained_variance_ratio_[1]*100:.1f}%)', fontsize=11)
plt.legend(loc='lower left', frameon=True, fontsize=9.5)
plt.tight_layout()
plt.savefig(figures_dir / 'pca_clusters_2019.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Lập bảng giá trị trung bình 6 yếu tố cơ sở (giữ nguyên tên cột FEATURE_COLUMNS)
profile_2019 = df_2019.groupby('cluster')[FEATURE_COLUMNS].mean()
profile_2019.index = [cluster_names[i] for i in profile_2019.index]
print('=== BẢNG PROFILE TRUNG BÌNH 6 ĐẶC TRƯNG CƠ SỞ NĂM 2019 ===')
display(profile_2019.round(4))

# Vẽ biểu đồ Radar Chart chân dung 3 cụm năm 2019 (nhãn cực giữ nguyên FEATURE_COLUMNS)
num_vars = len(FEATURE_COLUMNS)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8.5, 8.5), subplot_kw=dict(polar=True))
for idx in [0, 1, 2]:
    row_vals = df_2019[df_2019['cluster'] == idx][FEATURE_COLUMNS].mean().tolist()
    row_vals += row_vals[:1]
    ax.plot(angles, row_vals, linewidth=2.5, label=cluster_names[idx], color=palette_2019[idx])
    ax.fill(angles, row_vals, alpha=0.15, color=palette_2019[idx])

ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_ylim(0, 1.5)
radar_labels = ["gdp_per_capita", "social_support", "healthy_life_\nexpectancy", "freedom", "generosity", "corruption_\nperception"]
ax.set_thetagrids(np.degrees(angles[:-1]), radar_labels, fontsize=10, fontweight="bold")
ax.tick_params(pad=16)
ax.set_title('Chân Dung Radar 6 Yếu Tố Giữa 3 Cụm (Năm 2019)', y=1.12, fontsize=13, fontweight='bold')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.12), ncol=3, frameon=True, fontsize=9.5)
plt.tight_layout()
plt.savefig(figures_dir / 'radar_chart_2019.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Hậu kiểm phân bố điểm hạnh phúc (Happiness Score) thực tế năm 2019
plt.figure(figsize=(9, 5.5))
sns.boxplot(x='cluster_label', y='happiness_score', data=df_2019, palette=['#2ecc71', '#f39c12', '#e74c3c'], linewidth=1.5)
plt.title('Hậu Kiểm Độc Lập: Phân Bố Điểm Hạnh Phúc Giữa 3 Cụm (Năm 2019)', fontsize=13, pad=15, fontweight='bold')
plt.xlabel('Phân Cụm Tự Nhiên', fontsize=11)
plt.ylabel('Happiness Score (Thực Tế)', fontsize=11)
plt.xticks(fontsize=10)
plt.tight_layout()
plt.savefig(figures_dir / 'validation_happiness_score_2019.png', dpi=300, bbox_inches='tight')
plt.show()

print('Thống kê mô tả điểm hạnh phúc giữa 3 cụm năm 2019:')
stat_2019 = df_2019.groupby('cluster_label')['happiness_score'].agg(['count', 'mean', 'std', 'min', 'max'])
display(stat_2019.round(4))
```

## PHẦN B. MỞ RỘNG PHÂN TÍCH GOM CỤM CHUỖI THỜI GIAN 5 NĂM (2015–2019)
Để đánh giá tính bền vững và sự dịch chuyển cụm của các quốc gia, chúng tôi mở rộng phân tích gom cụm xuyên suốt 5 năm khảo sát (2015–2019).

```python
# Tải mô hình 5 năm và chuẩn hóa nhãn ngữ nghĩa
models = {y: json.load(open(project_root / f'models/kmeans_{y}.json', encoding='utf-8')) for y in years}
raw_dfs = {y: pd.read_csv(processed_dir / f'contrast_k3/contrast_k3_kmeans_{y}.csv') for y in years}

aligned_dfs = {}
for y in years:
    d = raw_dfs[y].copy()
    mean_scores = d.groupby('cluster_k3')['happiness_score'].mean().to_dict()
    sorted_clusters = sorted(mean_scores.keys(), key=lambda c: mean_scores[c], reverse=True)
    to_canonical = {orig: canonical for canonical, orig in enumerate(sorted_clusters)}
    d['canonical_cluster'] = d['cluster_k3'].map(to_canonical)
    aligned_dfs[y] = d

# Ghép cụm liên năm tham lam (Greedy Matching)
match_rows = []
for i in range(len(years) - 1):
    y_prev, y_curr = years[i], years[i+1]
    c_prev = models[y_prev]['model']['centroids']
    c_curr = models[y_curr]['model']['centroids']
    sc_prev = raw_dfs[y_prev].groupby('cluster_k3')['happiness_score'].mean().to_dict()
    sc_curr = raw_dfs[y_curr].groupby('cluster_k3')['happiness_score'].mean().to_dict()
    mapping = match_clusters_greedy(c_prev, c_curr, sc_prev, sc_curr)
    for id_curr, id_prev in mapping.items():
        dist = euclidean_distance(c_curr[id_curr], c_prev[id_prev])
        score_diff = abs(sc_curr[id_curr] - sc_prev[id_prev])
        match_rows.append({
            'base_year': y_prev,
            'target_year': y_curr,
            'base_cluster_raw': id_prev,
            'target_cluster_raw': id_curr,
            'base_happiness_mean': round(sc_prev[id_prev], 4),
            'target_happiness_mean': round(sc_curr[id_curr], 4),
            'centroid_euclidean_distance': round(dist, 4),
            'happiness_score_diff': round(score_diff, 4)
        })

df_matching = pd.DataFrame(match_rows)
df_matching.to_csv(tables_dir / 'cross_year_matching.csv', index=False)
print(f'Đã lưu bảng ghép cụm liên năm ({len(df_matching)} dòng) tại reports/tables/cross_year_matching.csv')
display(df_matching)
```

```python
# Lưới 2D PCA phân bố cụm 5 năm (pca_clusters_grid.png)
fig, axes = plt.subplots(1, 5, figsize=(22, 4.5), sharey=True)
palette_map = {0: '#2ecc71', 1: '#f39c12', 2: '#e74c3c'}

for idx, y in enumerate(years):
    sub_interim = df_all[df_all['year'] == y].dropna(subset=FEATURE_COLUMNS).reset_index(drop=True)
    X_mat = sub_interim[FEATURE_COLUMNS].values
    X_scaled = (X_mat - X_mat.mean(axis=0)) / X_mat.std(axis=0)
    pca = SimplePCA(n_components=2)
    X_pca = pca.fit_transform(X_scaled)
    
    merged_lbl = sub_interim.merge(aligned_dfs[y][['country', 'canonical_cluster']], on='country', how='left')
    lbls = merged_lbl['canonical_cluster'].fillna(1).astype(int).values
    
    ax = axes[idx]
    for c_id in [0, 1, 2]:
        mask = (lbls == c_id)
        ax.scatter(X_pca[mask, 0], X_pca[mask, 1], c=palette_map[c_id], label=f'Cụm {c_id+1}' if idx==0 else '', s=50, alpha=0.8)
    ax.set_title(f'Năm {y} (n={len(sub_interim)})', fontsize=12, fontweight='bold')
    ax.set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)', fontsize=10)
    if idx == 0:
        ax.set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)', fontsize=10)
    ax.grid(True, linestyle='--', alpha=0.5)

fig.suptitle('Lưới Phân Bố 3 Cụm Quốc Gia Trên Không Gian 2D PCA (2015–2019)', fontsize=15, y=1.03, fontweight='bold')
fig.legend(loc='upper center', bbox_to_anchor=(0.5, -0.06), ncol=3, frameon=True, fontsize=11)
plt.tight_layout()
plt.savefig(figures_dir / 'pca_clusters_grid.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Radar chart chân dung 3 cụm trung bình 5 năm (giữ nguyên FEATURE_COLUMNS)
all_aligned = pd.concat([aligned_dfs[y] for y in years], ignore_index=True)
mean_profiles_5y = df_all.merge(all_aligned[['country', 'year', 'canonical_cluster']], on=['country', 'year']).groupby('canonical_cluster')[FEATURE_COLUMNS].mean()

fig, ax = plt.subplots(figsize=(9, 9.2), subplot_kw=dict(polar=True))

for idx, row in mean_profiles_5y.iterrows():
    vals = row[FEATURE_COLUMNS].tolist()
    vals += vals[:1]
    ax.plot(angles, vals, linewidth=2.5, label=cluster_names[idx], color=palette_map[idx])
    ax.fill(angles, vals, alpha=0.15, color=palette_map[idx])

ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_ylim(0, 1.5)
radar_labels = ["gdp_per_capita", "social_support", "healthy_life_\nexpectancy", "freedom", "generosity", "corruption_\nperception"]
ax.set_thetagrids(np.degrees(angles[:-1]), radar_labels, fontsize=10, fontweight="bold")
ax.tick_params(pad=18)
ax.set_title('Chân Dung Radar 6 Yếu Tố Giữa 3 Cụm Quốc Gia (Trung Bình 5 Năm)', y=1.14, fontsize=14, fontweight='bold')
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.13), ncol=3, frameon=True, fontsize=10)
plt.tight_layout()
plt.savefig(figures_dir / 'radar_chart_clusters.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Biểu đồ xu hướng số lượng quốc gia theo từng cụm (cluster_count_trend.png)
trend_rows = []
for y in years:
    d = aligned_dfs[y].merge(df_all[df_all['year'] == y][['country'] + FEATURE_COLUMNS], on='country', how='left')
    for c_id in [0, 1, 2]:
        sub = d[d['canonical_cluster'] == c_id]
        trend_rows.append({
            'year': y,
            'cluster_id': c_id,
            'cluster_name': cluster_names[c_id],
            'country_count': len(sub),
            'mean_happiness_score': round(sub['happiness_score'].mean(), 4)
        })

df_trends = pd.DataFrame(trend_rows)
df_trends.to_csv(tables_dir / 'cluster_trends.csv', index=False)

plt.figure(figsize=(10, 6.2))
for c_id, color in [(0, '#2ecc71'), (1, '#f39c12'), (2, '#e74c3c')]:
    sub = df_trends[df_trends['cluster_id'] == c_id]
    plt.plot(sub['year'], sub['country_count'], marker='o', linewidth=2.5, markersize=8,
             label=cluster_names[c_id], color=color)
    for _, r in sub.iterrows():
        plt.text(r['year'], r['country_count'] + 0.8, f"{int(r['country_count'])}", ha='center', fontsize=10, fontweight='bold')

plt.title('Xu Hướng Số Lượng Quốc Gia Trong Từng Cụm (2015–2019)', fontsize=14, pad=35, fontweight='bold')
plt.xlabel('Năm Khảo Sát', fontsize=12)
plt.ylabel('Số Lượng Quốc Gia', fontsize=12)
plt.ylim(15, 95)
plt.xticks(years)
plt.legend(loc='lower center', bbox_to_anchor=(0.5, 1.02), ncol=3, frameon=True, fontsize=10)
plt.tight_layout()
plt.savefig(figures_dir / 'cluster_count_trend.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# Bảng dịch chuyển cụm 156 quốc gia và phân tích độ ổn định (country_cluster_shift.csv)
country_list = sorted(aligned_dfs[2019]['country'].unique())
shift_records = []
for c in country_list:
    row = {'country': c}
    c_labels = []
    for y in years:
        sub = aligned_dfs[y][aligned_dfs[y]['country'] == c]
        if len(sub) > 0:
            lbl = int(sub.iloc[0]['canonical_cluster'])
            row[f'cluster_{y}'] = lbl
            c_labels.append(lbl)
        else:
            row[f'cluster_{y}'] = np.nan
            c_labels.append(None)
    
    valid_labels = [l for l in c_labels if l is not None]
    if len(set(valid_labels)) == 1:
        status = 'Ổn định tuyệt đối'
    elif valid_labels[-1] < valid_labels[0]:
        status = 'Cải thiện nhóm (+)'
    elif valid_labels[-1] > valid_labels[0]:
        status = 'Suy giảm nhóm (-)'
    else:
        status = 'Biến động trung gian'
        
    row['shift_status'] = status
    row['is_stable'] = (len(set(valid_labels)) == 1)
    shift_records.append(row)

df_shift = pd.DataFrame(shift_records)
df_shift.to_csv(processed_dir / 'country_cluster_shift.csv', index=False)
print('=== PHÂN BỐ TRẠNG THÁI DỊCH CHUYỂN CỦA 156 QUỐC GIA (2015–2019) ===')
print(df_shift['shift_status'].value_counts())

print('\nDanh sách các quốc gia có sự dịch chuyển cụm:')
display(df_shift[~df_shift['is_stable']][['country', 'cluster_2015', 'cluster_2017', 'cluster_2019', 'shift_status']])
```

## PHẦN C. KẾT LUẬN, HẠN CHẾ VÀ HƯỚNG PHÁT TRIỂN (BƯỚC 11/11)

### 1. Kết luận chính từ thực nghiệm:
1. **Tính hợp lý của số cụm $k^*=3$:** Thuật toán K-Means và Hierarchical (Ward) from scratch đều đồng thuận chọn $k=3$ là số cụm tối ưu toàn cục xuyên suốt cả 5 năm (Silhouette cực đại trong khoảng $[0.293, 0.344]$).
2. **Sự phân hóa sâu sắc của các yếu tố cơ sở:**
   - **Cụm 1 (Phát triển toàn diện):** Chiếm ưu thế vượt trội ở GDP/người, Tuổi thọ và Hỗ trợ xã hội; đồng thời có chỉ số Nhận thức tham nhũng cao nhất (mức độ liêm chính cao).
   - **Cụm 2 (Đang phát triển):** Chiếm số lượng đông đảo nhất (79–85 quốc gia), có các chỉ số ở mức trung bình khá.
   - **Cụm 3 (Khó khăn & Thách thức):** Đối mặt với điểm số rất thấp ở hầu hết các trụ cột kinh tế - y tế - tự do.
3. **Độ ổn định cao qua chuỗi thời gian:** Hơn 91% quốc gia (142/156 nước) duy trì sự ổn định tuyệt đối trong cụm của mình; chỉ một số ít quốc gia có bước nhảy vọt (Benin, Ivory Coast) hoặc tụt hạng do khủng hoảng (Venezuela).
4. **Giá trị hậu kiểm thực chứng:** Mặc dù hoàn toàn không đưa `happiness_score` vào quá trình huấn luyện gom cụm, nhưng điểm hạnh phúc thực tế giữa 3 cụm có sự phân hóa cực kỳ sâu sắc và nhất quán qua cả 5 năm khảo sát.

### 2. Hạn chế của bộ dữ liệu và phương pháp:
- **Hạn chế dữ liệu:** Chỉ số hạnh phúc Cantril Ladder mang tính chất cảm nhận chủ quan; dữ liệu một số quốc gia bị khuyết chỉ số tham nhũng (như UAE năm 2018) đòi hỏi phải xử lý thận trọng.
- **Hạn chế thuật toán:** K-Means giả định cụm hình cầu lồi với phương sai đồng đều; DBSCAN gặp khó khăn khi mật độ dữ liệu toàn cầu thay đổi liên tục giữa các vùng địa lý.

- Triển khai mô hình trực quan tương tác phục vụ cộng đồng qua giao diện Streamlit tại `src/app/app.py`.
- Mở rộng kết hợp thêm các yếu tố vĩ mô khác như chỉ số bất bình đẳng thu nhập Gini, phát thải khí nhà kính và vị trí địa lý khu vực để phân nhóm chi tiết hơn.
