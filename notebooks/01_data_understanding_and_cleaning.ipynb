## [Bước 3/11] Đánh giá chất lượng dữ liệu (Data Quality Assessment)
Kiểm tra toàn diện các khía cạnh chất lượng dữ liệu qua 5 năm khảo sát (2015 - 2019):
1. **Sự bất đồng bộ cấu trúc và tên cột (Schema Inconsistency):**
   - Tên thuộc tính thay đổi liên tục qua các năm (ví dụ: `Happiness Rank` -> `Happiness.Rank` -> `Overall rank`).
   - Cột phân loại khu vực địa lý (`Region`) chỉ xuất hiện trong 2 năm đầu (2015 - 2016) và bị lược bỏ từ năm 2017 trở đi.
   - Sự hiện diện của các cột sai số đo lường khảo sát: `Standard Error` (2015), `Lower/Upper Confidence Interval` (2016), `Whisker.high/Whisker.low` (2017).
2. **Khảo sát giá trị thiếu (Missing values / NaN):**
   - Hầu hết các năm đều có dữ liệu đầy đủ cho các yếu tố cơ sở.
   - **Phát hiện ngoại lệ:** Năm 2018 bị khuyết thiếu đúng 1 giá trị tại thuộc tính `Perceptions of corruption` ở quốc gia **United Arab Emirates (UAE)**.
3. **Kiểm tra dữ liệu trùng lặp (Duplicate rows / Country names):**
   - Đảm bảo trong mỗi năm khảo sát không có quốc gia nào bị lặp lại dòng dữ liệu.

```python
# 1. Đánh giá tính nhất quán của tên cột
for year, df in raw_data.items():
    print(f'\n--- Danh sách cột năm {year} sau chuẩn hóa ---')
    print(list(df.columns))
```

```python
# 2. Kiểm tra giá trị thiếu (Missing values) trong từng năm
for year, df in raw_data.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    print(f'Năm {year}: {len(missing)} cột có giá trị thiếu')
    if len(missing) > 0:
        print(missing)
```

```python
# 3. Kiểm tra bản ghi trùng lặp (Duplicates)
for year, df in raw_data.items():
    dup_count = df.duplicated(subset=['country']).sum()
    print(f'Năm {year}: {dup_count} quốc gia bị trùng lặp')
```

## [Bước 4/11 - Phần 1] Tiền xử lý dữ liệu và Xây dựng tập dữ liệu trung gian (Interim Dataset)

Để chuẩn bị dữ liệu chuẩn xác cho các bước EDA và gom cụm tiếp theo, quy trình tiền xử lý tuân thủ các nguyên tắc khoa học sau:

### 1. Cơ sở khoa học của việc loại bỏ các cột sai số đo lường mẫu
- Các cột `Standard Error` (2015), `Lower Confidence Interval`, `Upper Confidence Interval` (2016), `Whisker.high`, `Whisker.low` (2017) là các chỉ số biểu thị sai số ước lượng mẫu và khoảng tin cậy 95% của cuộc thăm dò dư luận Gallup World Poll.
- Chúng phản ánh kích thước mẫu và độ biến động thống kê trong quy trình chọn mẫu, **hoàn toàn không mang giá trị nội hàm kinh tế - xã hội** cấu thành chất lượng sống của một quốc gia.
- Nếu giữ lại các cột này, chúng sẽ làm biến dạng không gian đặc trưng, gây nhiễu ma trận khoảng cách Euclidean và dẫn tới việc gom cụm bị lệch theo độ lớn sai số khảo sát thay vì theo bản chất phát triển quốc gia. Do đó, việc loại bỏ triệt để là bắt buộc.

### 2. Xử lý mẫu khuyết thiếu: Loại bỏ ngoại lệ United Arab Emirates (UAE) năm 2018
- Qua bước đánh giá chất lượng, quan sát duy nhất bị thiếu trong toàn bộ 5 năm là UAE năm 2018 tại cột `corruption_perception` (do cuộc khảo sát không thu thập được chỉ số này tại UAE năm 2018).
- Tỷ lệ thiếu chỉ là 1 / 782 mẫu (xấp xỉ 0.128%). Việc loại bỏ mẫu này (listwise deletion) là phương án tối ưu và an toàn, vì việc điền khuyết (imputation) bằng trung bình hoặc nội suy có thể tạo ra giả định sai lệch về nhận thức thể chế tại quốc gia Trung Đông này.
- Sau khi loại bỏ UAE 2018, tập dữ liệu còn lại chính xác **781 dòng**, bảo đảm 100% sạch (0 missing values trên 6 biến cơ sở).

### 3. Nguyên tắc bảo toàn bất biến của bài toán Gom cụm (Đề tài 16)
- **Ma trận đặc trưng gom cụm ($X$):** Tuyệt đối chỉ sử dụng 6 thuộc tính cơ sở:
  1. `gdp_per_capita` (Kinh tế)
  2. `social_support` (Hỗ trợ xã hội)
  3. `healthy_life_expectancy` (Tuổi thọ và sức khỏe)
  4. `freedom` (Tự do lựa chọn cuộc sống)
  5. `generosity` (Hào phóng / đóng góp cộng đồng)
  6. `corruption_perception` (Nhận thức tham nhũng / niềm tin thể chế)
- **Biến mục tiêu đối sánh:** `happiness_score` và `happiness_rank` được lưu trong bảng trung gian để đối chiếu, hậu kiểm ở Bước 9 (Notebook 04), tuyệt đối **không tham gia vào quá trình tính khoảng cách hay huấn luyện cụm**.

### 4. Bổ sung thông tin chuẩn hóa
- Điền khuyết cột `region` cho giai đoạn 2017-2019 dựa trên bảng tham chiếu 2015-2016 và ánh xạ địa lý.
- Bổ sung mã quốc tế `iso_alpha` (ISO 3166-1 alpha-3) cho tất cả các quốc gia để phục vụ trực quan hóa bản đồ không gian và ứng dụng tương tác.
- Giữ đúng cấu trúc 13 cột chuẩn theo `INTERIM_COLUMNS` và ghi ra tệp `data/interim/happiness_merged.csv`.

```python
interim_path = project_root / 'data' / 'interim' / 'happiness_merged.csv'
interim_df = build_interim_dataset(
    years=[2015, 2016, 2017, 2018, 2019],
    raw_dir=project_root / 'data' / 'raw',
    output_path=interim_path,
)

print(f'Đã sinh bảng dữ liệu trung gian chuẩn: {interim_df.shape} tại {interim_path}')
print('\nKiểm tra thông tin dữ liệu (interim_df.info()):')
interim_df.info()

print('\nKiểm tra số lượng giá trị thiếu trên từng cột:')
print(interim_df.isnull().sum())

core_features = [
    'gdp_per_capita',
    'social_support',
    'healthy_life_expectancy',
    'freedom',
    'corruption_perception',
    'generosity',
]
print(f'\nTổng số giá trị thiếu trong 6 thuộc tính cơ sở: {interim_df[core_features].isna().sum().sum()}')

uae_2018 = interim_df[(interim_df['year'] == 2018) & (interim_df['country'] == 'United Arab Emirates')]
print(f'Kiểm tra UAE 2018 tồn tại trong dữ liệu (kỳ vọng: 0 dòng): {len(uae_2018)} dòng')

display(interim_df.head())
```
